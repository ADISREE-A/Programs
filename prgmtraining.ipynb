1> Problem Statement :
An Autobiographical Number is a number N such that the first digit of N represents the count of how many zeroes are there in N, the second digit represents the count of how many ones are there in N and so on.
You are given a function, def FindAutoCount(n):
The function accepts string “n” which is a number and checks whether the number is an autobiographical number or not. If it is, an integer is returned, i.e. the count of distinct numbers in ‘n’. If not, it returns 0.
Assumption:
•	The input string will not be longer than 10 characters.
•	Input string will consist of numeric characters.
Note:
If string is None return 0.
Example:
Input:
n: “1210”
Output:
3
Explanation:
0th position in the input contains the number of 0 present in input, i.e. 1, in 1st position the count of number of 1s in input i.e. 2, in 2nd position the count of 2s in input i.e. 1, and in 3rd position the count of 3s i.e. 0, so the number is an autobiographical number.
Now unique numbers in the input are 0, 1, 2, so the count of unique numbers is 3. So 3 is returned.
```

The function accepts string str of size n as an argument. Implement the function which returns 1 if given string str is valid password else 0.
str is a valid password if it satisfies the below conditions.
•	– At least 4 characters
•	– At least one numeric digit
•	– At Least one Capital Letter
•	– Must not have space or slash (/)
•	– Starting character must not be a number
Assumption:
Input string will not be empty.
Example:
Input 1:
aA1_67
Input 2:
a987 abC012
Output 1:

Output 2:
0
```


````python
def CheckPassword(s,n):
    c,d,cap=0,0,0
    for ele in s:
        if ele.isalpha():
            c+=1
        elif ele.isupper():
            cap+=1
        elif ele.isdigit():
            d+=1
        elif ele==" " or ele=="/" or s[0].isdigit():
            return 0
    if c>=4 and d>=1 and cap>=1:
        return 1
s=input().split()
print(CheckPassword(s,len(s)))
````

````python
# Read input and handle multi-digit spacing constraints
n_input = input().strip().split()

if len(n_input) == 1:
    start = int(n_input[0])
    end = 1
else:
    start = int(n_input[0])
    end = int(n_input[1])

# Generate numbers in decreasing order
nums = list(range(start, end - 1, -1))

# Edge case: If there is only one number, it requires no operations
if len(nums) == 1:
    print(nums[0])
else:
    ops = []
    # Toggles to track alternation states (0 for first option, 1 for second option)
    even_toggle = 0  # 0: '/', 1: '*'
    odd_toggle = 0   # 0: '+', 1: '-'

    # Map operators dynamically based on the current number's parity
    for i in range(len(nums) - 1):
        if nums[i] % 2 == 0:
            if even_toggle == 0:
                ops.append('/')
            else:
                ops.append('*')
            even_toggle = 1 - even_toggle
        else:
            if odd_toggle == 0:
                ops.append('+')
            else:
                ops.append('-')
            odd_toggle = 1 - odd_toggle

    # Interleave numbers and operators into a list of tokens
    tokens = []
    for i in range(len(ops)):
        tokens.append(nums[i])
        tokens.append(ops[i])
    tokens.append(nums[-1])

    # Evaluate the expression by operation priority: "/", "*", "+", "-"
    for target_op in ['/', '*', '+', '-']:
        idx = 0
        while idx < len(tokens):
            if tokens[idx] == target_op:
                left = tokens[idx - 1]
                right = tokens[idx + 1]

                # Perform the operation
                if target_op == '/':
                    res = left // right  # Floor division
                elif target_op == '*':
                    res = left * right
                elif target_op == '+':
                    res = left + right
                elif target_op == '-':
                    res = left - right

                # Replace the three items (left, op, right) with the computed result
                tokens[idx - 1 : idx + 2] = [res]
            else:
                idx += 1

    # Output the final reduced value
    print(tokens[0])
````

````python
from itertools import combinations

def print_subsequences(arr):
    n = len(arr)
    print("sunsequence")
    for r in range(n + 1):
        for sub in combinations(arr, r):
            print(list(sub))

def print_subarrays(arr):
    n = len(arr)
    print("substring")
    for i in range(n):
        for j in range(i + 1, n + 1):
            print(arr[i:j])

arr = [1,2,3]
print_subsequences(arr)
print("\n")
print_subarrays(arr)
````
